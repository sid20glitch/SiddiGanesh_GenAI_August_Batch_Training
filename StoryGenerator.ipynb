```python
from google  import genai
from google.genai import types
from google.colab import userdata


client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
```

```python
for m in client.models.list():
  print(m.name)
```

```python
repsonse = client.models.generate_content(
    model = 'gemini-3.5-flash-lite',
    contents ='tell me one line of joke about developers.', #user prompts
    config=types.GenerateContentConfig(
        system_instruction='you are a helpful assistant',
        temperature=0.5,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(thinking_level='low')
        )
)


print("text generated by AI:",repsonse.text)

print()
u = repsonse.usage_metadata

print("usgae", repsonse.usage_metadata)



print("")

print("prompt_token_count", u.prompt_token_count)
print('candidates_token_count', u.candidates_token_count)

print('total_token_count', u.total_token_count)
print()
print()

```

```python
#build story generator and take inputs from users (topic, genre take from users)
from google  import genai
from google.genai import types
from google.colab import userdata


client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
MODEL = 'gemini-3.5-flash-lite'


def generate_story(topic, genre, length_word=200):

  prompt = f"""
    Write a {genre} story about {topic}.
    The story should be approximately {length_word} words.
    Give the story a title.
    Make it engaging with a clear beginning, middle, and ending.
    """

  response = client.models.generate_content(
        model=MODEL,
        contents=prompt,
        config=types.GenerateContentConfig(
            system_instruction="You are a creative story writer.",
            temperature=0.8,
            max_output_tokens=500
        )
    )

  return response.text



topic = input("Enter the topic: ")
genre = input("Enter the genre: ")
length = int(input("Enter story length: "))

story = generate_story(topic, genre, length)

print("\nGenerated Story:\n")
print(story)
```
